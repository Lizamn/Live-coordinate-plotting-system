```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import clear_output
import time

x_live = []
y_live = []

i = 0

# FIXED LIMITS (unknown initially)
x_limit = None
y_limit = None

while True:

    df = pd.read_excel("graphplot.xlsx")

    if i >= len(df):
        break

    x = df.iloc[i, 0]
    y = df.iloc[i, 1]

    x_live.append(x)
    y_live.append(y)

    # SET LIMITS ONLY ON FIRST POINT
    if x_limit is None:
        x_limit = max(df.iloc[:, 0]) + 2
        y_limit = max(df.iloc[:, 1]) + 2

    clear_output(wait=True)

    fig, ax = plt.subplots(figsize=(10, 6))

    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

    ax.scatter(x_live, y_live, s=250)

    for j in range(len(x_live)):
        ax.text(
            x_live[j] + 0.2,
            y_live[j] + 0.2,
            f"Z{j+1}"
        )

    # USE SAME LIMITS EVERY TIME
    ax.set_xlim(0, x_limit)
    ax.set_ylim(0, y_limit)

    ax.grid(alpha=0.3)

    ax.set_xlabel("X Axis")
    ax.set_ylabel("Y Axis")
    ax.set_title("Live Coordinate Plot")

    plt.show()

    i += 1
    time.sleep(1)
```
